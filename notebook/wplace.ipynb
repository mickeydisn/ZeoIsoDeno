```python
import requests
import time
import os
from datetime import datetime

# URL of the PNG file
url = "https://backend.wplace.live/files/s0/tiles/1037/704.png"


# Directory where images will be saved
save_dir = "images"
os.makedirs(save_dir, exist_ok=True)  # Create folder if it doesn’t exist


# Function to download and save the PNG with datetime in filename
def download_png():
    try:
        response = requests.get(url)
        response.raise_for_status()  # raise error if failed
        # Create filename with datetime
        filename = datetime.now().strftime("tile_%Y-%m-%d_%H-%M.png")
        filepath = os.path.join(save_dir, filename)
        with open(filepath, "wb") as f:
            f.write(response.content)
        print(f"Saved: {filename}")
    except Exception as e:
        print(f"Error: {e}")

# Run every 5 minutes
while True:
    download_png()
    time.sleep(300)  # 300 seconds = 5 minutes
```
