```python
from PIL import Image, ImageDraw
from PIL import Image, ImageDraw
import math

```

```python

def create_grid_image(
        nX, 
        nY, 
        cell_width=192, 
        cell_height=224, 
        border_width=1, 
        border_color=(0, 0, 0, 255), 
        tile_width=128, 
        tile_angle=30, 
        tile_fill=(255, 0, 0, 64), 
        tile_outline=(0, 0, 0, 255)
    ):  
    """
    Create a transparent image with a grid of rectangles with inner borders and isometric tiles.
    
    Parameters:
    - nX: Number of cells horizontally
    - nY: Number of cells vertically
    - cell_width: Width of each cell in pixels (default: 192)
    - cell_height: Height of each cell in pixels (default: 256)
    - border_width: Width of the border in pixels (default: 1)
    - border_color: RGBA color of the border (default: black opaque)
    - tile_width: Maximum width of the isometric tile (default: 128)
    - tile_angle: Angle for isometric projection in degrees (default: 30)
    - tile_fill: Fill color of the tile (default: light gray)
    - tile_outline: Outline color of the tile (default: black)
    
    Returns:
    - PIL Image object
    """
    # Calculate image dimensions
    width = nX * cell_width
    height = nY * cell_height
    
    # Create transparent image (RGBA mode)
    img = Image.new('RGBA', (width, height), (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)
    
    # Draw borders and isometric tiles for each cell
    for x in range(nX):
        for y in range(nY):
            # Calculate cell position
            left = x * cell_width
            top = y * cell_height
            right = left + cell_width
            bottom = top + cell_height
            
            # Draw inner border (rectangle outline)
            # We draw it slightly inset from the edges
            draw.rectangle(
                [left + border_width, top + border_width, 
                 right - border_width, bottom - border_width],
                outline=border_color,
                width=border_width
            )
            
            # Draw isometric tile at bottom center
            # Calculate tile dimensions using 30° angle
            # angle_rad = math.radians(tile_angle)
            
            # For an isometric diamond (losange) with 30° angle
            # Half width and height based on isometric projection
            half_width = tile_width / 2
            # half_height = half_width * math.tan(angle_rad)
            half_height = 32
            # Center position at bottom of cell
            center_x = left + cell_width / 2
            center_y = bottom - half_height - 6 # 10px padding from bottom
            
            # Define diamond points (top, right, bottom, left)
            diamond_points = [
                (center_x, center_y - half_height),           # Top
                (center_x + half_width, center_y),            # Right
                (center_x, center_y + half_height),           # Bottom
                (center_x - half_width, center_y)             # Left
            ]
            
            # Draw the isometric tile
            draw.polygon(diamond_points, fill=tile_fill, outline=tile_outline)
    
    return img

```

```python
# Create a 3x3 grid
nX = 4
nY = 20

grid_img = create_grid_image(nX, nY)

# Save the image
grid_img.save(f'grid_{nX}x{nY}.png')
print(f"Grid image created: {nX}x{nY} squares, total size: {grid_img.size}")

# Display in notebook (if running in Jupyter)
try:
    pass
    # from IPython.display import display
    # display(grid_img)
except:
    print("Not in Jupyter environment - image saved to file")
```
