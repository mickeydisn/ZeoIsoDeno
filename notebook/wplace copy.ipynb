```python
import os
import glob
import cv2
import numpy as np
from PIL import Image
from datetime import datetime
import random

tileX, tileY = "1037", "705"
# tileX, tileY = "1037", "705"

# --- CONFIG ---
BASE_IMAGE = f"images_save/{tileX}-{tileY}/tile_2025-08-27_00-29.webp"  # first image
DIFF_DIR = f"images_save_diff/{tileX}-{tileY}"        # directory containing diff_*.webp
OUTPUT_VIDEO = "output.mp4"
FPS = 40                  # video framerate
FRAME_DURATION = 180       # seconds per frame in simulated time

# --- Load base image ---
base_img = Image.open(BASE_IMAGE).convert("RGBA")
base_arr = np.array(base_img)

height, width = base_arr.shape[:2]

# --- Prepare video writer ---
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
video = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, FPS, (width, height))

# --- Load diff images ---
diff_files = sorted(glob.glob(os.path.join(DIFF_DIR, "diff_*.webp")))

def parse_timestamp(fname):
    # expects format: diff_%Y-%m-%d_%H-%M.webp
    ts = os.path.basename(fname)[5:-5]  # remove 'diff_' and '.webp'
    return datetime.strptime(ts, "%Y-%m-%d_%H-%M")

# Sort diffs by timestamp
diff_files = sorted(diff_files, key=parse_timestamp)

# Keep track of current working image
current_arr = base_arr.copy()

# Reference start time
start_time = parse_timestamp(diff_files[0])

print ("Image Diff Size - ", len(diff_files))
for diff_file in diff_files:
    print ("Image Diff ", diff_file)
    diff_img = Image.open(diff_file).convert("RGBA")
    diff_arr = np.array(diff_img)

    # Extract pixels that are different (non-transparent)
    mask = diff_arr[..., 3] > 0  # alpha channel > 0
    changed_pixels = np.argwhere(mask)

    # How many minutes since last?
    ts = parse_timestamp(diff_file)
    minutes = int((ts - start_time).total_seconds() // 60)
    start_time = ts
    # How many frames to spread?
    frames_needed = max(1, (minutes * 60) // FRAME_DURATION)

    print ("minutes:", minutes, frames_needed)

    # Shuffle changed pixels
    random.shuffle(changed_pixels)

    # Split pixels into groups
    groups = np.array_split(changed_pixels, frames_needed)

    for group in groups:
        frame = current_arr.copy()
        for y, x in group:
            frame[y, x] = diff_arr[y, x]
        current_arr = frame

        # Write frame multiple times to reach FPS
        bgr_frame = cv2.cvtColor(current_arr, cv2.COLOR_RGBA2BGR)
        video.write(bgr_frame)

# --- Release video ---
video.release()
print("Video created:", OUTPUT_VIDEO)
```

```python
import os
import glob
import cv2
import numpy as np
from PIL import Image
from datetime import datetime

tileX, tileY = "1037", "705"

# --- CONFIG ---
IMG_DIR = f"images_save/{tileX}-{tileY}"   # directory containing full frame images
OUTPUT_VIDEO = "output.mp4"
FPS = 20                   # video framerate
FRAME_DURATION = 60 * 5        # seconds per frame in simulated time

# --- Load all images ---
files = sorted(glob.glob(os.path.join(IMG_DIR, "tile_*.webp")))

def parse_timestamp(fname):
    # expects format: tile_%Y-%m-%d_%H-%M.webp
    ts = os.path.basename(fname)[5:-5]  # remove 'tile_' and '.webp'
    return datetime.strptime(ts, "%Y-%m-%d_%H-%M")

# Sort files by timestamp
files = sorted(files, key=parse_timestamp)

if not files:
    raise RuntimeError("No image files found in directory")

# Load first image to get dimensions
first_img = Image.open(files[0]).convert("RGBA")
width, height = first_img.size

# --- Prepare video writer ---
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
video = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, FPS, (width, height))

# Reference start time
start_time = parse_timestamp(files[0])

print("Total frames:", len(files))
for i, fname in enumerate(files):
    img = Image.open(fname).convert("RGBA")
    arr = np.array(img)

    ts = parse_timestamp(fname)
    minutes = int((ts - start_time).total_seconds() // 60)
    start_time = ts

    # How many frames to spread over simulated time?
    frames_needed = max(1, (minutes * 60) // FRAME_DURATION)
    print(f"{i+1}/{len(files)}: {fname} -> {minutes} minutes, {frames_needed} frames")

    bgr_frame = cv2.cvtColor(arr, cv2.COLOR_RGBA2BGR)

    # Write same frame multiple times to respect timing
    for _ in range(frames_needed):
        video.write(bgr_frame)

# --- Release video ---
video.release()
print("Video created:", OUTPUT_VIDEO)
```

```python
import os
import glob
import cv2
import numpy as np
from PIL import Image
from datetime import datetime

# --- CONFIG ---
TILES = [
    ("1036", "704"),
    ("1036", "705"),
    ("1037", "704"),
    ("1037", "705"),
]

BASE_DIR = "images_save"
OUTPUT_VIDEO = "output.mp4"
FPS = 20
FRAME_DURATION = 12060  # simulated seconds per frame

# --- Crop region (x1,y1) → (x2,y2) ---
CROP_X1, CROP_Y1 = 0, 0   # top-left corner
CROP_X2, CROP_Y2 = 2000, 2000  # bottom-right corner

# --- Collect all files from all tiles ---
tile_files = {}
for tx, ty in TILES:
    path = os.path.join(BASE_DIR, f"{tx}-{ty}")
    files = sorted(glob.glob(os.path.join(path, "tile_*.webp")))
    tile_files[(tx, ty)] = files

def parse_timestamp(fname):
    ts = os.path.basename(fname)[5:-5]  # remove 'tile_' and '.webp'
    return datetime.strptime(ts, "%Y-%m-%d_%H-%M")

# Build sorted list of all timestamps (union of all tiles)
all_timestamps = sorted({
    parse_timestamp(fname)
    for files in tile_files.values()
    for fname in files
})

if not all_timestamps:
    raise RuntimeError("No image files found in directories")

# Map (tile, timestamp) → filepath
tile_map = {}
for (tx, ty), files in tile_files.items():
    for fname in files:
        ts = parse_timestamp(fname)
        tile_map[(tx, ty, ts)] = fname

# --- Load first tile image to get tile size ---
sample_img = Image.open(next(iter(tile_files.values()))[0]).convert("RGBA")
tile_w, tile_h = sample_img.size

# Final canvas (before crop): 2x2 grid
full_w, full_h = tile_w * 2, tile_h * 2

# --- Prepare video writer with cropped size ---
crop_w = CROP_X2 - CROP_X1
crop_h = CROP_Y2 - CROP_Y1

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
video = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, FPS, (crop_w, crop_h))

# Reference start time
start_time = all_timestamps[0]

# Keep track of last available image per tile
last_images = {}

print("Total frames:", len(all_timestamps))
for i, ts in enumerate(all_timestamps):
    # Create empty canvas (RGBA)
    canvas = Image.new("RGBA", (full_w, full_h), (0, 0, 0, 255))

    # Place each tile in the right position
    positions = {
        ("1036", "704"): (0, 0),
        ("1037", "704"): (tile_w, 0),
        ("1036", "705"): (0, tile_h),
        ("1037", "705"): (tile_w, tile_h),
    }

    for (tx, ty), (x, y) in positions.items():
        fname = tile_map.get((tx, ty, ts))

        if fname:
            img = Image.open(fname).convert("RGBA")
            last_images[(tx, ty)] = img  # update last known
        else:
            # Use last known image if available, else fill with black
            img = last_images.get(
                (tx, ty),
                Image.new("RGBA", (tile_w, tile_h), (0, 0, 0, 255))
            )

        canvas.paste(img, (x, y))

    # Crop region
    cropped = canvas.crop((CROP_X1, CROP_Y1, CROP_X2, CROP_Y2))

    # How many minutes since last?
    minutes = int((ts - start_time).total_seconds() // 60)
    start_time = ts

    frames_needed = max(1, (minutes * 60) // FRAME_DURATION)
    print(f"{i+1}/{len(all_timestamps)}: {ts} -> {minutes} min, {frames_needed} frames")

    # Convert to BGR for OpenCV
    arr = np.array(cropped)
    bgr_frame = cv2.cvtColor(arr, cv2.COLOR_RGBA2BGR)

    # Write frame multiple times
    for _ in range(frames_needed):
        video.write(bgr_frame)

# --- Release video ---
video.release()
print("Video created:", OUTPUT_VIDEO)

```
